```python
from datetime import datetime,timedelta
class Account:
    def __init__(self,account_number,balance,*args,**kwargs):
        self.account_number=account_number
        self.balance=balance
        super().__init__(*args,**kwargs)
class LoggerMixin:
    def __init__(self,*args,**kwargs):
        print("Logger Initialized")
        super().__init__(*args, **kwargs) 
    def debit(self,amount):
            self.balance=1000
            if amount <= 0:print("Debit Amount must be positive")
            if amount > self.balance:print("Insufficient Balance")
            else:
                self.balance -= amount  
                current_dt = datetime.now() 
                print(f"Debited {amount}, After Debited: {self.balance} from {self.account_number} at {current_dt}")  
            super().debit(amount) if hasattr(super(), "debit")  else None   
    def credit(self,amount):
            self.balance=1000
            if amount <= 0:print("Credit amount must be positive")
            self.balance += amount
            next_current_dt = datetime.now() +timedelta(minutes=5)
            print(f"Credited {amount}, After Credited: {self.balance} from {self.account_number} at {next_current_dt}")
class AuditMixin:
     def __init__(self,*args,**kwargs):
          print("Audit Initialized")
          super().__init__(*args,**kwargs)
     def debit(self, amount):
        print(f"AUDIT: Debit {amount}")
        super().debit(amount) if hasattr(super(), "debit") else None
class BankAccount(AuditMixin, LoggerMixin, Account):
    def __init__(self, number, balance):
        super().__init__(number, balance)
A=BankAccount("ACC-1099",1000)
A.debit(200)
print("----")
A.credit(200) 

##hasattr(class,"attribute"):checks if the next class in the method resolution order (MRO) has a debit method or not 
# if the method has it will call the respective method otherwise skips without any error
```
