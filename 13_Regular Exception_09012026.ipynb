Q31. Parse a receipt line into named groups for banking case 1; use sample reference ACC-1031 and explain the result.

```python
Account_no="ACC-1031"
Info='''Balance:1000
Deposit:200
Withdraw:250
Date:19-06-2028'''
pattern=r"[A-Za-z]+:[\d\-\.]+"
result=re.findall(pattern,Info)
print(result)
dicton={}
for i,j in enumerate(result):
    i,j=j.split(":")
    dicton[i]=j
print(dicton)    
```

```python
import re

Account_no = "ACC-1031"
Info = '''Balance:1000
Deposit:200
Withdraw:250
Date:19-06-2028'''

# Regex with named groups for field and value
pattern = r"(?P<field>[A-Za-z]+):(?P<value>[\d\-\.]+)"

# Find all matches
matches = re.finditer(pattern, Info)

# Build dictionary directly from named groups
dicton = {}
for match in matches:
    dicton[match.group("field")] = match.group("value")

print(dicton)
```


Q80. Create a regex-based data quality report for banking case 1; use sample reference ACC-1080 and explain the result.

```python
import re

Account_no = "ACC-1080"

Data = '''ACC-1001 | Name: John Doe | Email: john.doe@example.com | Balance: 1500.00 | Currency: USD
ACC-1002 | Name: Priya Sharma | Email: priya.sharma@example | Balance: 2500.50 | Currency: INR
ACC-1003 | Name: Amit Kumar | Email: amit.kumar@example.com | Balance: 75,000 | Currency: EUR
ACC-1004 | Name: Jane Smith | Email: jane.smith@example.com | Balance: 1200.75 | Currency: GBP
ACC-1005 | Name: Rahul | Email: rahul@example.com | Balance: 150000 | Currency: JPY
ACC-1006 | Name: Maria Lopez | Email: maria.lopez@example.com | Balance: 999.99 | Currency: AUD
ACC-1007 | Name: Chen Wei | Email: chen.wei@example.com | Balance: 2345.67 | Currency: CAD
ACC-1008 | Name: Anna Müller | Email: anna.muller@example.com | Balance: 500.00 | Currency: CHF'''

# Regex patterns for validation
account = re.compile(r"^ACC-\d{4}$")
names = re.compile(r"^[A-Za-z\s]+$")
emails = re.compile(r"^[\w\.-]+@[\w\.-]+\.\w+$")
balance_pattern = re.compile(r"^\d{1,3}(,\d{3})*(\.\d{2})?$|^\d+(\.\d{2})?$")
currencies = re.compile(r"^[A-Z]{3}$")

# Split entries line by line
entries= Data.splitlines()

empty_report = []
for entry in entries:
    parts = [p.strip() for p in entry.split("|")]
    acc_id, name, email, balance, currency = parts

    acc_id = acc_id.strip()
    name = name.replace("Name:", "").strip()
    email = email.replace("Email:", "").strip()
    balance = balance.replace("Balance:", "").strip()
    currency = currency.replace("Currency:", "").strip()

    result = {
        "Account_ID": acc_id,
        "Account_Valid": bool(account.match(acc_id)),
        "Name": name,
        "Name_Valid": bool(names.match(name)),
        "Email": email,
        "Email_Valid": bool(emails.match(email)),
        "Balance": balance,
        "Balance_Valid": bool(balance_pattern.match(balance)),
        "Currency": currency,
        "Currency_Valid": bool(currencies.match(currency))
    }
    empty_report.append(result)

# Print Data Quality Report
for i in empty_report:
    print(i)
```


Q100. Create a regex-based data quality report for banking case 2; use sample reference ACC-1100 and explain the result.

```python
import re

Account_no = "ACC-1100"

Data = '''ACC-1001 | Name: John Doe | Email: john.doe@example.com | Balance: 1500.00 | Currency: USD
ACC-1002 | Name: Priya Sharma | Email: priya.sharma@example | Balance: 2500.50 | Currency: INR
ACC-1003 | Name: Amit Kumar | Email: amit.kumar@example.com | Balance: 75,000 | Currency: EUR
ACC-1004 | Name: Jane Smith | Email: jane.smith@example.com | Balance: 1200.75 | Currency: GBP
ACC-1005 | Name: Rahul | Email: rahul@example.com | Balance: 150000 | Currency: JPY
ACC-1006 | Name: Maria Lopez | Email: maria.lopez@example.com | Balance: 999.99 | Currency: AUD
ACC-1007 | Name: Chen Wei | Email: chen.wei@example.com | Balance: 2345.67 | Currency: CAD
ACC-1008 | Name: Anna Müller | Email: anna.muller@example.com | Balance: 500.00 | Currency: CHF'''

# Regex patterns for validation
account = re.compile(r"^ACC-\d{4}$")
names = re.compile(r"^[A-Za-z\s]+$")
emails = re.compile(r"^[\w\.-]+@[\w\.-]+\.\w+$")
balance_pattern = re.compile(r"^\d{1,3}(,\d{3})*(\.\d{2})?$|^\d+(\.\d{2})?$")
currencies = re.compile(r"^[A-Z]{3}$")

# Split entries line by line
entries= Data.splitlines()

empty_report = []
for entry in entries:
    parts = [p.strip() for p in entry.split("|")]
    acc_id, name, email, balance, currency = parts

    acc_id = acc_id.strip()
    name = name.replace("Name:", "").strip()
    email = email.replace("Email:", "").strip()
    balance = balance.replace("Balance:", "").strip()
    currency = currency.replace("Currency:", "").strip()

    result = {
        "Account_ID": acc_id,
        "Account_Valid": bool(account.match(acc_id)),
        "Name": name,
        "Name_Valid": bool(names.match(name)),
        "Email": email,
        "Email_Valid": bool(emails.match(email)),
        "Balance": balance,
        "Balance_Valid": bool(balance_pattern.match(balance)),
        "Currency": currency,
        "Currency_Valid": bool(currencies.match(currency))
    }
    empty_report.append(result)

# Print Data Quality Report
for i in empty_report:
    print(i)
```
