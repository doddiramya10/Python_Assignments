```python
#Calculator Using Functions
def calculator(operation):
    a=int(input("Enter the first number"))
    b=int(input("Enter the second number"))
    def add():
       return a+b
    def subtract():
        return a -b
    def multiply():
        return a*b
    def divide():
        if b ==0:
            return "Cannot divide by zero"
        else : 
         return a / b 

    if operation == "+":
        return add()  
    elif operation == "-":
        return subtract()
    elif operation == "*" :
        return multiply() 
    elif operation == "/" :
        return divide()  
    else:
        return "Invalid Operator"
operation = input("Enter the operator : +,-,*,/:")
result =calculator(operation) 
print("Result:",result)       
               
```

```python
#A function is a reusable block of code that performs a specific task. Functions improve code organization, reduce repetition, and make programs easier to maintain. In Python, functions are created using the def keyword.

# Example:

def display_employee():
    print("Employee details")
display_employee()
# In this project, functions are used to display employees, calculate salaries, generate reports, and handle menu operations.
```

```python
#2. Lambda Expressions

# A lambda expression is an anonymous function (a function without a name) used for short and simple operations. It can take multiple arguments but contains only one expression.

# Example:
lambda x: x["name"]
#In this project, lambda expressions are used with map(), filter(), reduce(), and sorted().
```

```python
# 3. map()

# The map() function applies a given function to every item in an iterable and returns a new iterable with the modified values.

# Syntax:

# map(function, iterable)

# Example:

employee_names = list(map(lambda x: x["name"], employees))

# In this project, map() is used to extract employee names and calculate revised salaries.
```

```python
#. filter()

# The filter() function is used to select elements from an iterable based on a condition. It returns only the elements where the condition evaluates to True.

# Syntax:

# filter(function, iterable)

# Example:

high_performers = list(filter(lambda x: x["performance_score"] >= 80, employees))

# In this project, filter() is used to find high-performing employees and search employees by department.
```

```python
# 6. Nested Functions

# A nested function is a function defined inside another function. The inner function can access variables from the outer function.

# Example:

def outer_function():
    def inner_function():
        print("Nested function")
    inner_function()

# In this project, nested functions are used inside generate_department_report() to calculate employee statistics.
```

```python
#A closure is a function that remembers variables from its enclosing function even after the outer function has finished execution.

# Example:

def create_bonus_calculator(percentage):
    def calculate(salary):
        return salary * percentage / 100
    return calculate

# In this project, closures are used to create different bonus calculators for 5%, 10%, and 15%.
```

```python
#LEGB Scope Explanation

# Python follows the LEGB rule to search for variables. The order is:

# Local → Enclosing → Global → Built-in
```

```python
#8. Local Scope

# A variable created inside a function is called a local variable. It can only be accessed within that function.

# Example:

def report():
    title = "Department Report"
    print(title)

# Here, title has local scope.
```

```python
#Enclosing Scope

# A variable created in an outer function and accessed by an inner function is called an enclosing variable.

# Example:

def report_function():
    department = "IT"

    def generate_report():
        print(department)

    generate_report()

# Here, department belongs to the enclosing scope.
```

```python
# # Global Scope

# A variable created outside all functions is called a global variable. It can be accessed throughout the program.

# Example:

company_name = "Edukron"

# In this project, company_name is a global variable accessed inside the report function.
```

```python
#  Built-in Scope

# Built-in scope contains Python's predefined functions and names that are always available.

# Examples:

# print()
# len()
# reduce()

# In this project, len(employees) uses the built-in len() function.
```
