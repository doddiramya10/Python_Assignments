```python
def find_min_max(*args): 
    if len(args) == 0: 
      return None, None 

    largest = max(args)
    smallest = min(args)


    return largest, smallest
Maximum, Minimum = find_min_max(50, 10, 80, 25, 5) 
print("Maximum:", Maximum) 
print("Minimum:", Minimum)
```

```python
def student_report(name, *marks, **kwargs): 
    total = sum(marks)
    average = sum(marks)/len(marks)

    passed = all(mark >= 35 for mark in marks)

    print("Student:", name)
    print("Course:", kwargs.get("course", "Not provided"))
    print("City:", kwargs.get("city", "Not provided"))
    print("Total:", total)
    print("Average:", average)

    if passed  :
       print("Result: PASS")
    else:
       print("Result: FAIL")
student_report( "Anil", 80, 75, 90, 65, course="Python", city="Hyderabad" )
```

```python
def calculate_salary(name, basic_salary, *allowance,**kwargs):
    allowance_total = sum(allowance)
    bonus = kwargs.get("bonus","value")
    tax_percentage = kwargs.get("tax", "value")

    gross_salary = basic_salary + allowance_total + bonus
    tax_amount = gross_salary * tax_percentage / 100
    net_salary = gross_salary - tax_amount

    print("Employee:", name)
    print("Gross salary:", gross_salary)
    print("Tax amount:", tax_amount)
    print("Net salary:", net_salary)


calculate_salary(
    "Rahul",
    50000,
    5000,
    3000,
    2000,
    bonus=10000,
    tax=10,
    department="IT"
)
```

```python
def count_data_types(*args):
    integer_count = 0
    string_count = 0
    list_count = 0
    other_count = 0

    for value in args:
        if type(value) is int:
            integer_count += 1
        elif isinstance(value,str ):
            string_count += 1
        elif isinstance(value, list):
            list_count += 1
        else:
            other_count += 1

    print("Integers:", integer_count)
    print("Strings:", string_count)
    print("Lists:", list_count)
    print("Others:", other_count)


count_data_types(10, "Python", [1, 2], 20, "SQL", 3.5, True)
```

```python
def separate_even_odd(*args):
    even_numbers = []
    odd_numbers = []

    for number in args:
        if number % 2 == 0:
              even_numbers.append(number)
        else:
            odd_numbers.append(number)

    return even_numbers, odd_numbers


evens, odds = separate_even_odd(10, 15, 22, 31, 40, 47)
print("Even numbers:", evens)
print("Odd numbers:", odds)
```

```python
def search_employees(*args, **kwargs):
    matched_employees = []

    for employee in args:
       matches = all(
            employee.get(key) == value
            for key, value in kwargs.items()
        )

       if matches:
            matched_employees.append(employee)

    return matched_employees


employee1 = {
    "name": "Rahul",
    "department": "IT",
    "city": "Bangalore"
}

employee2 = {
    "name": "Anil",
    "department": "HR",
    "city": "Hyderabad"
}

employee3 = {
    "name": "Priya",
    "department": "IT",
    "city": "Bangalore"
}

results = search_employees(
    employee1,
    employee2,
    employee3,
    department="IT",
    city="Bangalore"
)

print(results)
```

```python
def checkout(*args, **kwargs):
    subtotal = sum(args)

    discount_percentage = kwargs.get("discount", "value")
    tax_percentage = kwargs.get("tax", "value")

    discount_amount = subtotal * discount_percentage / 100
    discounted_amount = subtotal - discount_amount

    tax_amount = discounted_amount * tax_percentage / 100
    final_amount = discounted_amount + tax_amount

    print("Customer:", kwargs.get("customer", "Guest"))
    print("Subtotal:", round(subtotal, 2))
    print("Discount:", round(discount_amount, 2))
    print("Tax:", round(tax_amount, 2))
    print("Final amount:", round(final_amount, 2))


checkout(
    1000,
    500,
    250,
    customer="Rahul",
    discount=10,
    tax=18
)
```

```python
def register_user(**kwargs):
    required_fields = ["name", "email", "password"]

    missing_fields = [
        field
        for field in required_fields
        if field not in kwargs
    ]

    if  missing_fields:
        print("Missing fields:",missing_fields)
        return False

    print("Registration successful")
    return missing_fields


register_user(
    name="Rahul",
    email="rahul@example.com"
)
```

```python
def build_select_query(table, *columns, **filters):
    selected_columns = ", ".join(columns) if columns else filters

    query = f"SELECT {selected_columns} FROM {table}"

    if filters:
        conditions = []

        for key, value in filters.items():
            conditions.append(f"{key}='{value}'")

        query += " WHERE " + " AND ".join(conditions)

    return query


query = build_select_query(
    "Employee",
    "name",
    "salary",
    department="IT",
    city="Bangalore"
)

print(query)
```

```python
def log_function_call(func):

    def wrapper(*args,**kwargs):
        print("Function name:", func.__name__)
        print("Positional:", args)
        print("Keyword:", kwargs)
        print("Hi how are you?")

        result = func(*args, **kwargs)

        print("Result:", result)

        return result

    return wrapper


@log_function_call
def calculate_bill(price, quantity, discount=0):
    total = price * quantity
    discount_amount = total * discount / 100
    return total - discount_amount


bill=calculate_bill(500, 3, discount=10)
```

```python
def classify_numbers(*args,**kwargs):
    lower_limit = kwargs.get("lower_limit", "value")
    upper_limit = kwargs.get("upper_limit", "value")

    below = []
    within = []
    above = []

    for i in args:
        if i < lower_limit:
            below.append(i)
        elif i > upper_limit:
            above.append(i)
        else:
            within.append(i)

    return {
        "below": below,
        "within": within,
        "above": above
    }


classification = classify_numbers(
    5,
    15,
    25,
    35,
    45,
    lower_limit=20,
    upper_limit=40
)

print(classification)
```

```python
def merge_dictionaries(*dictionaries, **kwargs):
    merged = {}

    for dictionary in dictionaries:
        if isinstance(dictionary, dict):
            merged.update(dictionary)

    merged.update(**kwargs)

    return merged


result = merge_dictionaries(
    {"name": "Rahul", "city": "Bangalore"},
    {"department": "IT", "experience": 3},
    {"experience": 4},
    city="Hyderabad",
    salary=80000
)

print(result)
```

```python
import math
def etl_summary(source, destination, *tables, **kwargs):
    row_counts = kwargs.get("row_counts", {})
    batch_size = kwargs.get("batch_size", {})

    table_count = len(row_counts)

    total_rows = sum(
        row_counts.get(table, 0)
        for table in tables
    )

    batches = math.ceil(total_rows / batch_size ) if total_rows else 0

    print("Source:", source)
    print("Destination:", destination)
    print("Tables:", tables)
    print("Table count:", table_count)
    print("Total rows:", total_rows)
    print("Required batches:", batches)


etl_summary(
    "SQL Server",
    "ADLS Gen2",
    "Customer",
    "Orders",
    "Products",
    row_counts={
        "Customer": 12000,
        "Orders": 55000,
        "Products": 3000
    },
    batch_size=10000
)
```

```python
def check_permissions(*actions, **kwargs):
    role_permissions = {
        "admin": {"read", "write", "update", "delete"},
        "editor": {"read", "write", "update"},
        "viewer": {"read"}
    }

    role = kwargs.get("role", "value")
    allowed_actions = role_permissions.get(role, "value")

    print("User:", kwargs.get("username", "Unknown"))
    print("Role:", role)

    for action in actions:
        if action in allowed_actions:
            print(action, ": ALLOWED")
        else:
            print(action, ": DENIED")


check_permissions(
    "read",
    "update",
    "delete",
    username="Rahul",
    role="editor"
)
```

```python
def process_order(order_id, *products, **kwargs):
    subtotal = 0
    total_quantity = 0

    for product in products:
        if not isinstance(product, dict):
            continue

        price = product.get("price", "value")
        quantity = product.get("quantity", "value")

        product_total = price * quantity

        subtotal += product_total
        total_quantity += quantity

    discount_percentage = kwargs.get("discount", 0)
    tax_percentage = kwargs.get("tax", 0)

    discount_amount = subtotal * discount_percentage / 100
    amount_after_discount = subtotal - discount_amount

    tax_amount = amount_after_discount * tax_percentage / 100
    final_amount = amount_after_discount + tax_amount

    print("Order ID:", order_id)
    print("Customer:", kwargs.get("customer", "Guest"))
    print("Payment mode:", kwargs.get("payment_mode", "Not provided"))
    print("Total quantity:", total_quantity)
    print("Subtotal:", round(subtotal, 2))
    print("Discount amount:", round(discount_amount, 2))
    print("Tax amount:", round(tax_amount, 2))
    print("Final amount:", round(final_amount, 2))


process_order(
    "ORD-1001",
    {
        "name": "Laptop",
        "price": 50000,
        "quantity": 1
    },
    {
        "name": "Mouse",
        "price": 1000,
        "quantity": 2
    },
    {
        "name": "Keyboard",
        "price": 2000,
        "quantity": 1
    },
    customer="Rahul",
    payment_mode="UPI",
    discount=10,
    tax=18
)
```
