```python
# #Using filter() and reduce():

# Extract employees from the IT department.
# Calculate the total salary of IT employees.
employees = [
    {"id": 101, "name": "Rahul", "department": "IT", "salary": 45000},
    {"id": 102, "name": "Anil", "department": "HR", "salary": 65000},
    {"id": 103, "name": "Sneha", "department": "IT", "salary": 72000},
    {"id": 104, "name": "Kiran", "department": "IT", "salary": 82000},
    {"id": 105, "name": "Meena", "department": "Sales", "salary": 55000}
]

extract = list(filter(lambda x :x["department"]== "IT", employees))
total = reduce(lambda x,y : x + y["salary"],extract,0)
print(extract)
print(total)
```

```python
#Using filter() and reduce():

# Extract students who passed.
# Find the topper among the passed students.
students = [
    {"name": "Ravi", "marks": 92, "passed": True},
    {"name": "Kiran", "marks": 68, "passed": True},
    {"name": "Meena", "marks": 85, "passed": True},
    {"name": "Ajay", "marks": 48, "passed": True},
    {"name": "Sneha", "marks": 35, "passed": False},
    {"name": "Rahul", "marks": 95, "passed": True}
]

extract = list(filter(lambda x : x["marks"] if x["passed"]== True else False,students))
topper = reduce (lambda x,y :  y["marks"] if x < y["marks"] else x , extract,0)

print(extract)
print(topper)
```

```python
#Using filter(), map(), and reduce():

# Extract students scoring more than 70.
# Extract only their marks.
# Calculate the average marks.
students = [
    {"name": "Ravi", "marks": 92},
    {"name": "Kiran", "marks": 68},
    {"name": "Meena", "marks": 85},
    {"name": "Ajay", "marks": 48},
    {"name": "Sneha", "marks": 79},
    {"name": "Rahul", "marks": 95}
]
extract = list(filter(lambda x : x["marks"] > 70 ,students))
mark = list(map(lambda x :x["marks"],extract ))
total = reduce(lambda x,y : x+y , mark )
average = total/len(mark)
print(extract)
print(mark)
print(total)
print(average)
```

```python
#Using filter(), map(), and reduce():

# Extract products currently in stock.
# Calculate each product's inventory value using price × stock.
# Calculate the total inventory value.
products = [
    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 0},
    {"name": "Monitor", "price": 18000, "stock": 8},
    {"name": "Printer", "price": 12000, "stock": 0},
    {"name": "SSD", "price": 6500, "stock": 12}
]

extract = list(filter(lambda x:x if x["stock"]!=0 else False, products))
calculate= list(map(lambda x : x["price"]*x["stock"],extract))
total = reduce(lambda x,y : x+y,calculate)
print(extract)
print(calculate)
print(total)
```

```python
#Using filter(), map(), and reduce():

# Extract products belonging to the Electronics category.
# Apply a 20% discount.
# Calculate the total discounted value.

products = [
    {"name": "Laptop", "price": 55000, "category": "Electronics"},
    {"name": "Chair", "price": 4500, "category": "Furniture"},
    {"name": "Monitor", "price": 18000, "category": "Electronics"},
    {"name": "Desk", "price": 8500, "category": "Furniture"},
    {"name": "SSD", "price": 6500, "category": "Electronics"}
]

extract = list(filter(lambda x : x["category"]=="Electronics" ,products))
discounted = list(map(lambda x : x["price"]* 0.80,extract))
total = reduce(lambda x,y : x + y ,discounted)
print(extract)
print(discounted)
print(total)
```

```python
#Using filter() and reduce():

# Extract delivered orders.
# Calculate the total delivered revenue.
orders = [
    {"order_id": 1001, "amount": 1200, "status": "Delivered"},
    {"order_id": 1002, "amount": 3500, "status": "Pending"},
    {"order_id": 1003, "amount": 8500, "status": "Delivered"},
    {"order_id": 1004, "amount": 4200, "status": "Cancelled"},
    {"order_id": 1005, "amount": 7600, "status": "Delivered"},
    {"order_id": 1006, "amount": 5100, "status": "Pending"}
]

extract = list(filter(lambda x : x["status"]== "Delivered",orders))
revenue = reduce(lambda x,y : x+y["amount"],extract,0)
print(extract)
print(revenue)
```

```python
#Using filter(), map(), and reduce():

# Extract transactions whose amount is greater than 5000.
# Add 18% GST to every selected transaction.
# Calculate the grand total after GST.
transactions = [
    {"id": 1, "amount": 1200},
    {"id": 2, "amount": 5200},
    {"id": 3, "amount": 18000},
    {"id": 4, "amount": 2500},
    {"id": 5, "amount": 9600},
    {"id": 6, "amount": 7200}
]

extract = list(filter(lambda x : x["amount"] > 5000,transactions))
GST = list(map(lambda x : x["amount"]* 1.18,extract))
calculate = reduce(lambda x ,y : x+y,GST)
print(extract)
print(GST)
print(calculate)
```

```python
#Using filter(), map(), and reduce():

# Extract active salespeople from the South region.
# Calculate a 5% incentive for each selected salesperson.
# Create a list containing salesperson name, sales amount, and incentive.
# Calculate the total incentive payable.

sales_records = [
    {"salesperson": "Rahul", "region": "South", "sales": 45000, "active": True},
    {"salesperson": "Anil", "region": "North", "sales": 65000, "active": True},
    {"salesperson": "Sneha", "region": "South", "sales": 72000, "active": False},
    {"salesperson": "Kiran", "region": "South", "sales": 82000, "active": True},
    {"salesperson": "Meena", "region": "West", "sales": 55000, "active": True},
    {"salesperson": "Arjun", "region": "South", "sales": 98000, "active": True}
]

extract = list(filter(lambda x: x["active"]==True and x["region"]== "South",sales_records))
increase = list(map(lambda x :{"name":x["salesperson"],"sales":x["sales"],"incentive": x["sales"]*0.05},extract))
create = reduce(lambda x,y : x+y["incentive"],increase,0)

print(extract)
print(increase)
print(create)
```
